# Guesser evaluation environment check

This notebook verifies the `uv` environment and the cross-platform image/scientific Python stack. It does **not** evaluate a model or report recognition results.

From the repository root, start Jupyter with:

```sh
uv sync --project research/guesser-evaluation --locked
uv run --project research/guesser-evaluation jupyter lab research/guesser-evaluation/notebooks
```

````python
from importlib.metadata import version
import platform
import sys

import matplotlib
import numpy as np
import pandas as pd
from PIL import Image

import guesser_evaluation

assert sys.version_info[:2] == (3, 12), sys.version
image = Image.new("L", (224, 224), color=255)
pixels = np.asarray(image)
assert pixels.shape == (224, 224)
assert pd.DataFrame({"pixel": pixels[0, :2]}).shape == (2, 1)

environment = {
    "python": sys.version.split()[0],
    "executable": sys.executable,
    "os": platform.system(),
    "architecture": platform.machine(),
    "workspace_package": guesser_evaluation.__file__,
    "packages": {
        name: version(name)
        for name in ("ipykernel", "jupyterlab", "matplotlib", "numpy", "pandas", "pillow")
    },
}
environment
````
